```sql
USE CATALOG vd_dab_try;

CREATE SCHEMA IF NOT EXISTS gold;
```

```sql
CREATE OR REPLACE TABLE gold.trips_agg AS
SELECT
  pickup_zip,
  dropoff_zip,
  COUNT(*) AS trip_count,
  AVG(trip_distance) AS avg_trip_distance,
  SUM(fare_amount) AS total_fare
FROM silver.trips
GROUP BY pickup_zip, dropoff_zip;
```
